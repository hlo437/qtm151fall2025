## Quadratic Model + Correct Plot

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Example data (replace x_data and y_data with quiz names)
x = np.array([1, 2, 3, 4, 5])
y = np.array([2.1, 2.9, 7.2, 12.4, 18.9])

# Fit y ≈ a x^2 + b x + c
coeffs = np.polyfit(x, y, 2)
a, b, c = coeffs
print("Quadratic coefficients:", coeffs)

# Model curve
x_line = np.linspace(x.min(), x.max(), 200)
y_line = a*x_line**2 + b*x_line + c

# Plot
plt.figure()
plt.scatter(x, y, label="data")
plt.plot(x_line, y_line, label="model")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

## Currency CSV + Percentage Growth

```python
import pandas as pd
import matplotlib.pyplot as plt

# Replace with quiz filename
df = pd.read_csv("currency_rates.csv")

# Replace column names if quiz uses different ones
# Example columns:
#   "date", "EUR_USD", "GBP_USD"

df["eur_growth"] = df["EUR_USD"].pct_change() * 100
df["gbp_growth"] = df["GBP_USD"].pct_change() * 100

plt.figure()
plt.plot(df["date"], df["eur_growth"], label="EUR growth (%)")
plt.plot(df["date"], df["gbp_growth"], label="GBP growth (%)")
plt.xlabel("date")
plt.ylabel("percent growth")
plt.legend()
plt.show()
```

## Boolean Column: Conversion Rate > Given Value

```python
threshold_value = 1.10   # replace with quiz value

df["eur_above_threshold"] = df["EUR_USD"] > threshold_value
print(df[["date", "EUR_USD", "eur_above_threshold"]])
```

## fill_between Highlighting High-Rate Regions

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(df["date"], df["EUR_USD"], label="EUR to USD")

plt.fill_between(
    df["date"],
    df["EUR_USD"],
    where=df["eur_above_threshold"],
    alpha=0.3
)

plt.xlabel("date")
plt.ylabel("EUR to USD")
plt.legend()
plt.show()
```

## Sin/Cos DataFrame + diff() Derivative Column

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Example: 200 points from 0 to 2π
t = np.linspace(0, 2*np.pi, 200)

df = pd.DataFrame({
    "time": t,
    "sin": np.sin(t),
    "cos": np.cos(t)
})

# Derivative of sin using diff()
df["d_sin_dt"] = df["sin"].diff() / df["time"].diff()

plt.figure()
plt.plot(df["time"], df["cos"], label="cos(t)")
plt.plot(df["time"], df["d_sin_dt"], label="d(sin)/dt")
plt.xlabel("time")
plt.legend()
plt.show()
```
